## 2. O método dos mínimos quadrados (MMQ)

### Princípios fundamentais do MMQ

- Desenvolvido por **Gauss** e **Legendre** (1795–1810)
- Partindo-se de um sistema redundante de equações lineares  $A\vec{x} = \vec{l}$, 
    - onde $\vec{l}$ é o vetor das observações eivadas de erros $\vec{v}$, $A$ é a matriz de coeficientes e $\vec{x}$ é o vetor de parâmetros
- Modelo linear: $A\vec{x_a} = \vec{l} + \vec{v}$, onde $\vec{x_a}$ é o vetor de parâmetros ajustados
- Minimiza a **soma dos quadrados dos resíduos**: $\sum v_i^2 = \vec{v}^T \vec{v} = ||\vec{v}||_2 = \min$
- Solução: $\vec{x_a} = (A^T A)^{-1} A^T \vec{l}$
- Após a solução, o menor vetor de resíduos é $\vec{v}= A\vec{x_a} - \vec{l}$

**Propriedades do estimador MMQ:**

- Solução única do sistema indeterminado se `posto(A) = nr de incógitas`
- Estimador imparcial (não tendencioso)
- Solução de mínima variância
- Solução de máxima verossimilhança (erros normais)

**Exercício 05 (resolvido):** Dado o sistema de equações redundante abaixo, qual a solução através do MMQ?

$$ \begin{cases}
m = 5.11 \\
m = 4.97 \\
m = 5.05 \\
m = 4.95
\end{cases} 
$$

**Solução exercício 05:**

Em forma matricial $A\vec{x} = \vec{l}$:

$
\underbrace{\begin{bmatrix}
1 \\ 1 \\ 1 \\ 1
\end{bmatrix}}_{A} \underbrace{[m]}_{\vec{x}} = 
\underbrace{\begin{bmatrix}
5.11 \\ 4.97 \\ 5.05 \\ 4.95
\end{bmatrix}}_{\vec{l}}
$

Cálculo da aplicação da fórmula:

````python
import sympy as sp

# Matriz A e vetor l_vec
A = sp.Matrix([[1],
               [1],
               [1],
               [1]])
l_vec = sp.Matrix([[5.11], 
                   [4.97], 
                   [5.05], 
                   [4.95]])

# Solução
x_a = (A.T * A).inv() * A.T * l_vec

print("Vetor solução x_a:")
x_a
````

#### Complemento: Demonstração da fórmula do ajustamento MMQ

$ A\vec{x}=\vec{l} \rightarrow A\vec{x_a}=\vec{l_a}=\vec{l}+\vec{v}$,

onde $A$ é a matriz design, $\vec{x}$ é vetor de parâmetros a serem ajustados, $\vec{l}$ é o vetor de observações, $\vec{x_a}$ é o vetor de parâmetros ajustados, $\vec{l_a}$ é o vetor das observações ajustados e $\vec{v}$ é o vetor de resíduos a serem minimizados.

Se aprofundando na forma do sistema linear, considerando $m$ observações e $n$ incógitas/parâmetros, tem-se que:

$\underbrace{\underbrace{[A]_{m \times n} [\vec{x_a}]_{n \times 1} = [\vec{l}]_{m \times 1}}_{\text{sistema inconsistente}} + [\vec{v}]_{m \times 1}}_{\text{sistema indeterminado}}$ 

cuja solução é única, se A de posto completo, i.e., `posto(A)=n`.

Somatório do quadrado os resíduos deve ser minimizado:

$\vec{v}^T\vec{v} = (A\vec{x_a} - \vec{l})^T(A\vec{x_a} - \vec{l})$
$=(\vec{x_a}^TA^T - \vec{l}^T)(A\vec{x_a} - \vec{l})$
$=\vec{x_a}^TA^TA\vec{x_a}  - \vec{x_a}^TA^T\vec{l} -  \vec{l}^TA\vec{x_a} + \vec{l}^T\vec{l} = mín$

> Nota: cada parcela da equação acima é um escalar, portanto equivale dizer $\vec{x_a}^TA^T\vec{l} = (\vec{x_a}^TA^T\vec{l})^T = \vec{l}^TA\vec{x_a}$.

Derivada com respeito a $\vec{x}$ no ponto em que os parâmetros estão ajustados $\vec{x_a}$ igualada a zero:

$ \frac{\partial v^Tv}{\partial \vec{x}}  \bigg|_{\vec{x}=\vec{x_a}} = 0 $
$ 2A^TA\vec{x_a} - 2A^T\vec{l} = 0$
$ \vec{x_a} = (A^TA)^{-1} A^T \vec{l} $

Resíduos:

$ \vec{v} = A\vec{x_a} - \vec{l}  $

Erro aleatório total:

$ e_a = \vec{v}^T\vec{v}  $

Observações ajustadas:

$ \vec{l_a}=\vec{l} + \vec{v} $

## Complemento: Matricialização de problemas

Feitas $m$ obsevações de $n$ variáveis aleatórias:

$\begin{cases}x = l_x^1 \\ \vdots  \\ x = l_x^m\end{cases}$ &emsp;&emsp;
... &emsp;&emsp;
$\begin{cases}z = l_z^1 \\ \vdots  \\ z = l_z^m\end{cases}$

De onde se abstrai vários sistemas lineares com soluções calculadas separadamente:

$
\vec{l_x} = \begin{bmatrix} l_x^1 \\ \vdots \\ l_z^1 \end{bmatrix}_{m \times 1},
\vec{x} = \begin{bmatrix} x \end{bmatrix},
A_{x} = \begin{bmatrix} 1 \\\vdots \\ 1  \end{bmatrix}_{m \times 1} \rightarrow 
A_{x}\vec{x} = \vec{l_x}\\ \vdots \\
$
$
\vec{l_z} = \begin{bmatrix} l_z^1 \\ \vdots \\ l_z^1 \end{bmatrix}_{m \times 1},

\vec{x} = \begin{bmatrix} z \end{bmatrix},
A_{z} = \begin{bmatrix} 1 \\ \vdots \\ 1  \end{bmatrix}_{m \times 1}\rightarrow 
A_{z}\vec{x} = \vec{l_z}
$

A notação algébrica matricial do modelo linear pode ser juntada em um único problema da seguinte forma:

O vetor de todas as observações é empilhado em um único vetor-coluna:

$
L'_{n\times m} = \begin{bmatrix} \vec{l_x}^T \\ \vdots \\  \vec{l_z}^T\end{bmatrix} = 
\begin{bmatrix} l_x^1 & ... & l_x^m \\ \vdots & \ddots & \vdots \\ l_z^1 & ... & l_z^m \end{bmatrix}
\rightarrow
L_{n m \times 1} = \begin{bmatrix} \vec{l_x} \\ \vdots \\  \vec{l_z}\end{bmatrix} = 
\begin{bmatrix} l_x^1 \\ \vdots \\ l_x^m \\ \vdots \\ l_z^1 \\ \vdots \\ l_z^m \end{bmatrix}
$

O vetor de todos os prâmetros é empilhado em um único vetor-coluna:

$
\vec{x} = X_{n \times 1} = 
\begin{bmatrix} x \\ \vdots \\ z  \end{bmatrix}
$

Neste caso, a matriz de coeficiente é diagonal (depende do problema):

$A_{n\times m} = \begin{bmatrix} [1]_{m \times 1} &  &  0 \\ & \ddots & \\ 0 & & [1]_{m \times 1} \end{bmatrix}$

Assim, fazendo a manipulação matricial acima, podemos aplicar a solução MMQ diretamente a:

$A X = L$

> Nota: Perceba que a matriz de coeficientes $A$ precisa ser *desenhada* engenhosamente para se encaixar no problema, ela também é conhecida por matriz *design* devido a esta característica. 

> Nota: Nas próximas aulas e daqui para frente, o vetor de parâmetros $\vec{x}$ e de observações $\vec{l}$ passarão a ser chamados matrizes coluna $X$ e $L$ para corresponder a uma abordagem mais genérica.

## Complemento: Demonstração da fórmula do MMQ para modelos não lineares

Os problemas não lineares, em geral, se encaixam na seguinte forma:

$F(X_a) = F(X_0 + \Delta X) = L_a = L_b + V $, onde:
- $F$ é uma função não linear;
- $L_b$ é o vetor de observações com erros;
- $X_a$ é o vetor de incógnitas ajustadas;
- $L_a$ é o vetor de observações ajustadas;
- $V$ é o vetor de resíduos;
- $X_0$ é vetor com valores aproximados dos parâmetros, converge para $X_a$ após algumas iterações;
- $\Delta X$ é o Vetor correção (incremento) dentro de uma iteração, converge para $0$ após algumas iterações;

Neste caso, expandindo o termo $L_b + V = F(X_0 + X)$ pela série de Taylor e desprezando os termos maiores que a 2ª ordem, tem-se:

$L_b + V = F(X_0 + X) ≈ \underbrace{F(X_0)}_{L_0}+ \underbrace{\frac{dF}{dX} \bigg\rvert_{X=X_0}}_{A}  (X_a-X_0) $

$\Phi = V^TV = (\underbrace{L_0-L_b}_{L}+A\underbrace{(X_a-X_0)}_{\Delta X})^T(\underbrace{L_0-L_b}_{L}+A\underbrace{(X_a-X_0)}_{\Delta X}) = (A\Delta X+L)^T(A\Delta X+L) = min  $

$\frac{d\Phi}{dX} \bigg\rvert_{X=\Delta X} = 2A^TA\Delta X + 2A^TL = 0$

$\Delta X = -(A^TA)^{-1}A^TL$

- $A$ é a matriz das derivadas parcias da função não linear em relação às incógnitas (parâmetros);
- $L_0$ é a função dos parâmetros aproximados, converge para $L_a$ após algumas iterações;;
- $V$ é o vetor de resíduos;
- $V=A\Delta X+L$ é o modelo linearizado do método paramétrico.

O resultado do Vetor correção ($\Delta X$) é utilizado para atualizar o valor de $X_0$ e $L_0$ para a iteração seguinte:

$X^{(2)}_0 = \Delta X^{(1)} + X_0^{(1)} $

$L^{(2)}_0 = A\Delta X^{(1)} + L_0^{(1)}$

$L^{(2)} = L^{(2)}_0 - L_b $

$A^{(2)} = \frac{dF}{dX} \bigg\rvert_{X=X^{(2)}_0}$

$\Delta X^{(2)} = -({A^{(2)}}^TA^{(2)})^{-1}{A^{(2)}}^TL^{(2)} $

E assim até que $X^{(i)}$ se reduza a 0 com determinado grau de tolerânica arbitrário.

**Exercício 13 (resolvido)**: Encontre a solução via MMQ do seguinte sistema não linear:

$$
\begin{split}
f_1&: \ x+y-2y^2 =-4   \\
f_2&: \ x^2+y^2 = 8   \\
f_3&: \ 3x^2-y^2 =7.7
\end{split}
$$

**Solução exercício 13:**

$X_a = \begin{bmatrix} x \\  y \end{bmatrix}$

$
F(X_a) = \begin{bmatrix} f_1(X_a) \\  f_2(X_a) \\ f_3(X_a) \end{bmatrix} = 
\begin{bmatrix}
    x+y-2y^2\\
    x^2+y^2 \\
    3x^2-y^2
\end{bmatrix}$

$
L_b+V = \begin{bmatrix}
-4 + v_1 \\
8 + v_2 \\
7.7 + v_3
\end{bmatrix}
$

Jacobiana:

$
A =
\begin{bmatrix}
\frac{df_1}{dX_a} \\
\frac{df_2}{dX_a} \\
\frac{df_3}{dX_a}
\end{bmatrix} =
\begin{bmatrix}

  \frac{\partial f_1}{\partial x} & \frac{\partial f_1}{\partial y} \\
  \frac{\partial f_2}{\partial x} & \frac{\partial f_2}{\partial y} \\
  \frac{\partial f_3}{\partial x} & \frac{\partial f_3}{\partial y}
\end{bmatrix}
$

Arbitrando $X^{(1)}_0 = [1 \ 1]$ (cuidado para não obter $A$ singular durante o chute!), obtemos $L^{(1)}_0 = F(X^{(1)}_0)$ para a 1ª iteração. devemos encontrar $\Delta X^{(1)}$ através do MMQ não linear, repetindo os cálculos até a tolerância desejada.

Implementação:

````python
from sympy import symbols, Matrix, diff
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

# Definindo as variáveis simbólicas
x, y = symbols('x y')

# funções
f1 = x+y-2*y**2 
f2 = x**2+y**2 
f3 = 3*x**2-y**2
F = Matrix([[f1],
            [f2],
            [f3]])

# Derivadas parciais
# Montando a matriz Jacobiana
A = Matrix([[diff(f1, x), diff(f1, y)],
            [diff(f2, x), diff(f2, y)],
            [diff(f3, x), diff(f3, y)]]) 
print("Matriz Jacobiana:")
display(A)

# mount L
Lb = np.array([[-4],
               [8],
               [7.7]])

# Inicialização chute e correção (X0 e X)
X_0 = {x: 1, y: 1} 
print("X0:\n", np.array(list(X_0.values())))

### Início do ajustametno 
erros = [] # guardar cada erro durante as iterações para compor um gráfico
erro = 10000000
while erro>10e-6:
    # update L0 = F(X0)
    L0 = np.array(F.subs(X_0).evalf().tolist(), dtype=float)
    # update L
    L = L0 - Lb
    # update A
    a = np.array(A.subs(X_0).evalf().tolist(), dtype=float)
    # update X = vetor das correções
    DeltaX = -np.linalg.inv(a.T @ a) @ a.T @ L
    DeltaX = DeltaX.reshape(-1,)
    print("DeltaX:\n", DeltaX)
    # update erro
    erro = np.linalg.norm(DeltaX)
    erros.append(erro)
    print("Erro:\n", erro)
    # update X0
    X0 =  np.array(list(X_0.values()))
    X0 = DeltaX + X0
    print("\nXa:\n", X0)
    X_0 = {x: X0[0], y: X0[1]} 

# Gráfico
plt.title("Convergência do erro no ajustamento")
plt.xlabel("Iterações")
plt.ylabel("Erro")
plt.grid()
plt.plot(erros)
plt.show()
````
