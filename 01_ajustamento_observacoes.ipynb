## 2. Conceituar o ajustamento das observações

**Ajustamento** é o ramo da matemática aplicada que busca uma <b style="color:#2ECC40">solução aproximada (idealmente única)</b> para <b  style="color:#ffff00">sistemas inconsistentes</b>, admitindo resíduos nas observações e estimando a qualidade da solução obtida.

Em termos práticos, ajustar significa **acomodar um conjunto de observações por meio de um modelo matemático**, de modo que o vetor solução represente a **melhor estimativa** para os parâmetros desconhecidos.

## 3. Conceito de medidas, observações e modelos matemáticos

### **3.1 Medidas**

**Medida** é uma operação física de comparação com um padrão. O resultado desta medida (leitura) é a observação. Exemplos: Medidas de ângulos, de distâncias, da temperatura... etc.

<img src=media/imgs/medidas.png width=800>

- **Medida direta**: direta sobre grandeza procurada. Ex: distância ou ângulo isolado
- **Medida indireta**: cálculo das incógnitas (parâmetros) por meio de relações matemáticas de medidas diretas. Os erros de medida nas observações são propagados para os parâmetros. Ex: área e coordenadas.

### **3.3 Modelos matemáticos**

**Modelo matemático** é conjunto de relações que representa o fenômeno observado. Pode ser:

- **Modelo determinístico/funcional**: descreve as relações determinísticas, ou seja, completamente previsíveis, entre observações e parâmetros. Por exemplo, a soma dos ângulos internos de um triângulo ($ABC$) possui uma modelagem funcional: $A+B+C=180^\circ$ em que, <b style="color:#2ECC40">dadas as mesmas entradas, o resultado será sempre o mesmo</b>;
- **Não determinístico/estocástico** : descreve o comportamento aleatório do evento através de suas dispersões e variâncias. Por exemplo, as inúmeras medições da massa do pacote de arroz constrói uma amostragem cujo modelo estocástico é equivalente à distribuição gaussiana. Em suma, o <b style="color:#2ECC40">resultado pode variar mesmo com condições parecidas</b>;
- **Modelo explícito**: a variável de interesse está <b style="color:#2ECC40">isolada</b> diretamente na equação;
- **Modelo implícito**: a variável de interesse aparece <b style="color:#2ECC40">misturada</b> em uma equação a ser resolvida.

## 4. Identificar os princípios fundamentais do método dos mínimos quadrados (MMQ).

- Desenvolvido por **Gauss** e **Legendre** (1795–1810)
- Partindo-se de um sistema redundante de equações lineares  $A\vec{x} = \vec{l}$, 
    - onde $\vec{l}$ é o vetor das observações eivadas de erros $\vec{v}$, $A$ é a matriz de coeficientes e $\vec{x}$ é o vetor de parâmetros
- Modelo linear: $A\vec{x_a} = \vec{l} + \vec{v}$, onde $\vec{x_a}$ é o vetor de parâmetros ajustados
- Minimiza a **soma dos quadrados dos resíduos**: $\sum v_i^2 = \vec{v}^T \vec{v} = ||\vec{v}||_2 = \min$
- Solução: $\vec{x_a} = (A^T A)^{-1} A^T \vec{l}$
- Após a solução, o menor vetor de resíduos é $\vec{v}= A\vec{x_a} - \vec{l}$

### **4.1 Propriedades do estimador MMQ:**

- Solução única do sistema indeterminado se `posto(A) = nr de incógnitas`
- Estimador imparcial (não tendencioso)
- Solução de mínima variância
- Solução de máxima verossimilhança (erros normais)

**Exercício 05 (resolvido):** Dado o sistema de equações redundante abaixo, qual a solução via MMQ?

$$ \begin{cases}
m = 5.11 \\
m = 4.97 \\
m = 5.05 \\
m = 4.95
\end{cases} 
$$

**Solução exercício 05:**

Em forma matricial $A\vec{x} = \vec{l}$:

$
\underbrace{\begin{bmatrix}
1 \\ 1 \\ 1 \\ 1
\end{bmatrix}}_{A} \underbrace{[m]}_{\vec{x}} = 
\underbrace{\begin{bmatrix}
5.11 \\ 4.97 \\ 5.05 \\ 4.95
\end{bmatrix}}_{\vec{l}}
$

Modelo linear $A\vec{x_a} = \vec{l} + \vec{v}$, onde $\vec{x_a}$ é o vetor de parâmetros ajustados

$
\underbrace{\begin{bmatrix}
1 \\ 1 \\ 1 \\ 1
\end{bmatrix}}_{A} \underbrace{[m]}_{\vec{x_a}} = 
\underbrace{\begin{bmatrix}
5.11 \\ 4.97 \\ 5.05 \\ 4.95
\end{bmatrix}+\begin{bmatrix}
v_1 \\ v_2 \\ v_3 \\ v_4
\end{bmatrix}}_{\vec{l_a}}
$

$||\vec{v}|| = \min \rightarrow \vec{x_a} = (A^T A)^{-1} A^T \vec{l}$

```python
import sympy as sp
from IPython.display import display

# Matriz A e vetor l_vec
A = sp.Matrix([[1],
               [1],
               [1],
               [1]])
l_vec = sp.Matrix([[5.11], 
                   [4.97], 
                   [5.05], 
                   [4.95]])

# Solução
x_a = (A.T * A).inv() * A.T * l_vec
v = A * x_a - l_vec
print(f"Vetor solução x_a=")
display(x_a)
print(f"Vetor resíduos v=")
display(v)
```
