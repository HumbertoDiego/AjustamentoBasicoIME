#### Conceito de ajustamento das observações

**Ajustamento** é o ramo da matemática aplicada que busca uma solução aproximada — idealmente única — para sistemas inconsistentes, admitindo resíduos nas observações e estimando a qualidade da solução obtida.

Em termos práticos, ajustar significa **acomodar um conjunto de observações por meio de um modelo matemático**, de modo que o vetor solução represente a **melhor estimativa** para os parâmetros desconhecidos.

#### Conceito de medidas, observações e modelos matemáticos

**Medida**: operação física de comparação com um padrão. O resultado desta medida (leitura) é a observação. Exemplos: Medidas de ângulos, de distâncias, da temperatura... etc.
- **Medida direta**: direta sobre grandeza procurada. Ex: distância ou ângulo isolado
- **Medida indireta**: cálculo das incógnitas (parâmetros) por meio de relações matemáticas de medidas diretas. Os erros de medida nas observações são propagados para os parâmetros. Ex: área e coordenadas.

**Observação**: resultado numérico da medida (a leitura), já afetado por flutuações probabilísticas. Contém erros que decorrem não apenas de **falhas humanas**, mas também da **imperfeição do equipamento**; da influência das **condições ambientais** nas quais se processa a
mensuração, e outros **efeitos aleatórios** sem causa específica. São medidas de qualidade das observações a precisão e a acurácia, vistos posteriomente.

**Modelo matemático**: conjunto de relações que representa o fenômeno observado.

- **modelo funcional**: descreve as relações determinísticas, ou seja, completamente previsíveis, entre observações e parâmetros. Por exemplo, a soma dos ângulos internos de um triângulo ($ABC$) possui uma modelagem funcional: $A+B+C=180^\circ$
- **modelo estocástico**: descreve o comportamento probabilístico das observações, ou seja, enquadra a aleatoriedade do evento através de suas dispersões e variâncias. Por exemplo, as inúmeros medições da massa do pacote de arroz constrói uma amostragem cujo modelo estocástico é equivalente à distribuição gaussiana.

> Nota: Relembre que a distribuição gaussiana, ou normal, é uma distribuição de probabilidade contínua em forma de sino, definida pela média $\mu$ e pelo desvio padrão $\sigma$, amplamente utilizada para modelar fenômenos naturais e erros de medição. Nela, dada a distribuição de uma amostra, infere-se que dentre $\mu \pm 1\sigma$ se encontra 68% da população, dentre $2\sigma$, 95% e dentre $3\sigma$, 99,7% da população (Regra 68-95-99.7).

### 2. O método dos mínimos quadrados (MMQ)

#### Princípios fundamentais do MMQ

- Desenvolvido por **Gauss** e **Legendre** (1795–1810)
- Partindo-se de um sistema redundante de equações lineares  $A\vec{x} = \vec{l}$, 
    - onde $\vec{l}$ é o vetor das observações eivadas de erros $\vec{v}$, $A$ é a matriz de coeficientes e $\vec{x}$ é o vetor de parâmetros
- Modelo linear: $A\vec{x_a} = \vec{l} + \vec{v}$, onde $\vec{x_a}$ é o vetor de parâmetros ajustados
- Minimiza a **soma dos quadrados dos resíduos**: $\sum v_i^2 = \vec{v}^T \vec{v} = ||\vec{v}||_2 = \min$
- Solução: $\vec{x_a} = (A^T A)^{-1} A^T \vec{l}$
- Após a solução, o menor vetor de resíduos é $\vec{v}= A\vec{x_a} - \vec{l}$

**Propriedades do estimador MMQ:**

- Solução única do sistema indeterminado se `posto(A) = nr de incógitas`
- Estimador imparcial (não tendencioso)
- Solução de mínima variância
- Solução de máxima verossimilhança (erros normais)

**Exercício 05 (resolvido):** Dado o sistema de equações redundante abaixo, qual a solução através do MMQ?

$$
 \begin{cases}
m = 5.11 \\
m = 4.97 \\
m = 5.05 \\
m = 4.95
\end{cases} 
$$

**Solução exercício 05:**

Em forma matricial $A\vec{x} = \vec{l}$:

$$
\underbrace{\begin{bmatrix}
1 \\ 1 \\ 1 \\ 1
\end{bmatrix}}_{A} \underbrace{[m]}_{\vec{x}} = 
\underbrace{\begin{bmatrix}
5.11 \\ 4.97 \\ 5.05 \\ 4.95
\end{bmatrix}}_{\vec{l}}
$$

Cálculo da aplicação da fórmula:

````python
import sympy as sp

# Matriz A e vetor l_vec
A = sp.Matrix([[1],
               [1],
               [1],
               [1]])
l_vec = sp.Matrix([[5.11], 
                   [4.97], 
                   [5.05], 
                   [4.95]])

# Solução
x_a = (A.T * A).inv() * A.T * l_vec

print("Vetor solução x_a:")
x_a
````

## UD II

### 1. Conceito de erro e sua natureza. Classificação dos erros: grosseiros, sistemáticos e acidentais.

Se os erros de medida não existissem, não haveria necessidade de ajustamento... seriam sistemas com solução única resolvidos com recursos usuais de álgebra linear. 

O Ajustamento visa encontrar uma solução aproximada (preferencialmente única e melhor) para sistemas de equações inconsistentes e estimar a sua
precisão, admitindo-se a existência de resíduos nas observações.

Porém existem outras formas de minimizar os erros nas observações. Formas práticas, ligadas à natureza de cada tipo de erro.

Como visto anteriormente, os erros nas observações podem ser originados por **falhas humanas**, **imperfeição do equipamento**, **condições ambientais** e **efeitos aleatórios** sem causa específica. Derivada dessas naturezas, os erros são classificados como.

- **Erros Grosseiros**: grandes, detectáveis e podem ser totalmente eliminados
- **Erros Sistemáticos**: seguem lei conhecida, podendo ser reduzidos ou eliminados
- **Erros Acidentais (aleatórios)**: pequenos, imprevisíveis, distribuição normal (base do MMQ)

Perceba que o método MMQ é uma etapa de um processo maior. Dada uma observação $l$, se soubéssemos seu valor real $l_r$ teríamos que descontar todos os tipos de erro:

$$l = l_r - e_g - e_s - e_a$$

Onde $e_g$, $e_s$ e $e_a$ são os erros grosseiros, sistemáticos e aleatórios.

### 2. Estudo experimental dos erros acidentais. Experiência de Bradley. Conceito de incerteza e exatidão.


#### A experiência de Bradley.


James Bradley (1693–1762) foi um astrônomo inglês que, ao tentar medir a paralaxe estelar (o deslocamento aparente das estrelas causado pelo movimento da Terra), fez medições repetidas e sistemáticas da posição de uma mesma estrela ao longo do tempo. Ao registrar centenas de medições, Bradley percebeu que:

- Os resultados variavam aleatoriamente em torno de um valor central
- Os erros não eram sistemáticos, pois não havia tendência constante para um lado
- Ao calcular a média de todas as medições, o resultado era muito mais confiável do que qualquer medição isolada
- A média das medições converge para um valor
- Quanto mais medições, menor a dispersão da média, ou seja, maior confiabilidade

$$

#### Os conceitos de incerteza e exatidão

Como falado anteriormente as medidas de qualidade de uma observação são a precisão e a acurácia:

- Precisão (ou incerteza): dispersão (desvio padrão). O grau de precisão obtido é dependente da qualidade do equipamento que se faz as medidas, da estabilidade ambiental e da habilidade do operador em controlar os procedimentos de medição

<img src="media/imgs/img4.png" width=600>

- Acurácia (ou exatidão): proximidade do valor verdadeiro. Normalmente expressa pela diferença absoluta entre uma observação e seu valor verdadeiro. Na prática, partindo-se do princípio que uma quantidade verdadeira nunca pode ser determinada, a acurácia sempre será **uma incógnita**. 

> Nota: Na literatura, aceita-se que a Acurácia engloba também a Precisão. Logo, valores imprecisos não são acurados.

### 3. Erros verdadeiros e aparentes. Resíduos.

Sendo $l$ o valor observado, $l_v$ o valor verdadeiro (desconhecido na prática) de uma observação e $l_a$ o seu valor ajustado, para cada observação, tem-se:

- **Erro verdadeiro:** &emsp;  $|l - l_v|$ (desconhecido)
- **Erro aparente:** &emsp;&emsp;  $|l - l_a|$
- **Resíduo ($v$):** &emsp;&emsp;&emsp; $|l_a - l|$ (a correção necessária para tornar o sistema compatível com o modelo ajustado)

> Nota: No MMQ, os resíduos são escolhidos de modo a minimizar o valor do escalar $(v^T v)$, ou seja, minimizar a Perda Quadrática Média (MSE). Porém, nada impede de escolher outras funções de custo mais complexas. Assunto explorado em aprendizado profundo (*deep learning*).


#### Complemento: Demonstração da fórmula do ajustamento MMQ

$ A\vec{x}=\vec{l} \rightarrow A\vec{x_a}=\vec{l_a}=\vec{l}+\vec{v}$,

onde $A$ é a matriz design, $\vec{x}$ é vetor de parâmetros a serem ajsutados, $\vec{l}$ é o vetor de observações, $\vec{x_a}$ é o vetor de parâmetros ajustados, $\vec{l_a}$ é o vetor das observações ajustados e $\vec{v}$ é o vetor de resíduos a serem minimizados.

Se aprofundando na forma do sistema linear, considerando $m$ observações e $n$ incógitas/parâmetros, tem-se que:

$\underbrace{\underbrace{[A]_{m \times n} [\vec{x_a}]_{n \times 1} = [\vec{l}]_{m \times 1}}_{\text{sistema inconsistente}} + [\vec{v}]_{m \times 1}}_{\text{sistema indeterminado}}$ 

cuja solução é única, se A de posto completo, i.e., `posto(A)=n`.

Somatório do quadrado os resíduos deve ser minimizado:

$$\vec{v}^T\vec{v} = (A\vec{x_a} - \vec{l})^T(A\vec{x_a} - \vec{l})$$
$$=(\vec{x_a}^TA^T - \vec{l}^T)(A\vec{x_a} - \vec{l})$$
$$=\vec{x_a}^TA^TA\vec{x_a}  - \vec{x_a}^TA^T\vec{l} -  \vec{l}^TA\vec{x_a} + \vec{l}^T\vec{l} = mín$$

> Nota: cada parcela da equação acima é um escalar, portanto equivale dizer $\vec{x_a}^TA^T\vec{l} = (\vec{x_a}^TA^T\vec{l})^T = \vec{l}^TA\vec{x_a}$.

Derivada com respeito a $\vec{x}$ no ponto em que os parâmetros estão ajustados $\vec{x_a}$ igualada a zero:
        
$$ \frac{\partial v^Tv}{\partial \vec{x}}  \bigg|_{\vec{x}=\vec{x_a}} = 0 $$
$$ 2A^TA\vec{x_a} - 2A^T\vec{l} = 0$$
$$ \vec{x_a} = (A^TA)^{-1} A^T \vec{l} $$

Resíduos:

$$ \vec{v} = A\vec{x_a} - \vec{l}  $$

Erro aleatório total:

$$ e_a = \vec{v}^T\vec{v}  $$

Observações ajustadas:

$$ \vec{l_a}=\vec{l} + \vec{v} $$

#### Complemento: Matricialização de problemas

Feitas $m$ obsevações de $n$ variáveis aleatórias:

$\begin{cases}x = l_x^1 \\\vdots  \\x = l_x^m\end{cases}$ &emsp;&emsp;
... &emsp;&emsp;
$\begin{cases}z = l_z^1 \\\vdots  \\z = l_z^m\end{cases}$

De onde se abstrai vários sistemas lineares com soluções calculadas separadamente:
$$
\vec{l_x} = \begin{bmatrix} l_x^1 \\ \vdots \\ l_z^1 \end{bmatrix}_{m \times 1},
\vec{x} = \begin{bmatrix} x \end{bmatrix},
A_{x} = \begin{bmatrix} 1 \\\vdots \\ 1  \end{bmatrix}_{m \times 1} \rightarrow 
A_{x}\vec{x} = \vec{l_x}\\ \vdots \\
$$
$$
\vec{l_z} = \begin{bmatrix} l_z^1 \\ \vdots \\ l_z^1 \end{bmatrix}_{m \times 1},
\vec{x} = \begin{bmatrix} z \end{bmatrix},
A_{z} = \begin{bmatrix} 1 \\\vdots \\ 1  \end{bmatrix}_{m \times 1}\rightarrow 
A_{z}\vec{x} = \vec{l_z}
$$

A notação algébrica matricial do modelo linear pode ser juntada em um único problema da seguinte forma:

$$
L'_{n\times m} = \begin{bmatrix} \vec{l_x}^T \\ \vdots \\  \vec{l_z}^T\end{bmatrix} = 
\begin{bmatrix} l_x^1 & ... & l_x^m \\ \vdots & \ddots & \vdots \\ l_z^1 & ... & l_z^m \end{bmatrix}
\rightarrow
L_{n m \times 1} = \begin{bmatrix} \vec{l_x} \\ \vdots \\  \vec{l_z}\end{bmatrix} = 
\begin{bmatrix} l_x^1 \\ \vdots \\ l_x^m \\ \vdots \\ l_z^1 \\ \vdots \\ l_z^m \end{bmatrix}
$$

$$
\vec{x} = X_{n \times 1} = 
\begin{bmatrix} x \\ \vdots \\ z  \end{bmatrix}
$$

$$
A_{n\times m} = \begin{bmatrix} [1]_{m \times 1} &  &  0 \\ & \ddots & \\ 0 & & [1]_{m \times 1} \end{bmatrix}
$$

$$
A X = L
$$

> Nota: Perceba que a matriz de coeficientes $A$ precisa ser *desenhada* engenhosamente para se encaixar no problema, ela também é conhecida por matriz *design* devido a esta característica. 

> Nota: A partir das próximas aulas, o vetor de parâmetros $\vec{x}$ e de observações $\vec{l}$ passarão a ser chamados matrizes coluna $X$ e $L$ para corresponder a uma aboradegem mais genérica.
