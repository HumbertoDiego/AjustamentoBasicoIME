## O Problema

Existe uma classe de problemas em ajustamento em que certas **condições** precisam ser **exatamente satisfeitas**.

**Exemplo:** Em redes de triangulação, satisfazer a condição geométrica $S = (n-2)180°$.

**Exemplo:** Além de uma minimização dos resíduos, satisfazer a condição de que a norma das observações seja unitária.


## 1. Apresentar o Modelo Condicionado

No **Modelo Condicionado**, não se estimam parâmetros diretamente no ajustamento. O objetivo é corrigir as observações para que elas satisfaçam <b style="color:#2ECC40">um conjunto de equações de condição independentes</b> e representadas genericamente por:

$$
b_{11}(\ell_1+v_1)
+
b_{12}(\ell_2+v_2)
+\cdots+
b_{1n}(\ell_r+v_r)
-
c_1=0
$$

$$
b_{21}(\ell_1+v_1)
+
b_{22}(\ell_2+v_2)
+\cdots+
b_{2n}(\ell_r+v_r)
-
c_2=0
$$

$$
\vdots
$$

$$
b_{m1}(\ell_1+v_1)
+
b_{m2}(\ell_2+v_2)
+\cdots+
b_{mn}(\ell_r+v_r)
-
c_m =0
$$

onde $b$ são os coeficientes das observações $l$, $v$ os respectivos resíduos desconhecidos, $l+v=\hat{l}$ representa as observações ajustadas, enquanto os termos $c$ são constantes inerentes às condições, $m$ é a quantidade de equações de condição e $r$ é a quantidade de valores observados.

### **1.1 Comparação Paramétrico - Condicionado**

**Exemplo 03:** Quatro observações $a_1, a_2, b$ e $c$ em cima da ângulos de um triângulo: 

$\text{Paramétrico: } \begin{cases} x_1 = a_1 \\ x_1 = a_2 \\x_2 = b\\ 180 -x_1-x_2 = c\end{cases} \rightarrow gl=4eq -2 par = 2 \rightarrow \qquad  \begin{cases} x_1 = a_1 +v_1 \\ x_1 = a_2  +v_2\\x_2 = b +v_3\\ 180 -  x_1 -  x_2 = c +v_4\end{cases}  $

$\text{Condicionado: } \begin{cases} a_1 +b +c - 180 =0 \\ a_1-a_2=0\end{cases}  \rightarrow gl=2eq -0 par = 2 \rightarrow  \begin{cases} (a_1 +v_1)+ (b +v_2) +(c +v_3) - 180=0\\ (a_1+v_1)-(a_2+v_2)=0\end{cases} $

### **1.2 Em termos matriciais para o caso linear**

$$
F(\hat{L}) = F(L+V)= B (L+V) - {}_mC_1 =0
$$

$$
BV + \underbrace{BL - C}_{-W} = 0 
$$

$$
{}_mB_r \  {}_rV_1  - {}_mW_1  =0
$$

onde:
- $L = [l_1 \ \ ... \ \ l_r]^T$ é o vetor de observações (anteriormente $L_b$);
- $\hat{L} = [\hat{l}_1 \ \ ... \ \ \hat{l}_r]^T$ é o vetor de observações ajustadas (anteriormente $L_a$);
- $V = [v_1 \ \ ... \ \ v_r]^T$ é o vetor dos respectivos resíduos;
- $C = [c_1 \ \ ... \ \ c_m]^T$ é o vetor das constantes das condições;
- $B$ é a matriz dos coeficiente também chamada de matriz design $(m \times r)$, por não ser simétrica, impossibilita o isolamento de $V$; e
- $W = C -BL$ é chamado de vetor dos **erros de fechamento** das condições, isto é, o quanto as observações brutas deixam de satisfazer as condições.

### **1.3 Para o caso não linear**

As condições não podem ser escritas diretamente como $BV=W$. Para tanto, utilizamos suas aproximações linearizadas pela expansão de Taylor truncada na 1ª ordem:


$$F(\hat{L}) = F(L + V) ≈ \underbrace{F(L)}_{-W}+ \underbrace{\frac{\partial F}{\partial\hat{L}} \bigg\rvert_{\hat{L}=L}}_{B}  V =0$$

$$
{}_mB_r \ {}_rV_1  - {}_mW_1 =0
$$

onde:
- $B$ é a matriz das derivadas parciais (jacobiana) das equações de condição em relação às observações; e
- $W=F(L)$ é o vetor dos fechamentos.

### **1.4 Ponderação das observações**

Cada resíduo está associado a uma observação específica $(v_i​=\hat{l}_i​−l_i)$​, deseja-se que uma observação muito precisa tenha peso grande e uma observação pouco precisa tenha peso pequeno, isto é traduzido ao montar a função de custo (ou função de perda ou função objeto):

$$ \Phi =V^TPV=p_1​v_1^2​+p_2​v_2^2​+\dots+p_r​v_r^2​ $$

onde $p_i$ é o peso associado à observação $l_i$, frequentemente o inverso do desvio padrão da observação.


## 2. Apresentar a solução para a modelagem Condicionada

Deseja-se minimizar o somatório dos quadrados ponderados dos resíduos sujeito às equações de condição. Em termos algébricos:

$$ \argmin_{V}(V^TPV) \qquad\text{ sujeito a }\qquad BV - W =0 $$

Problemas de minimização sujeita a condição são resolvidos classicamente pela técnica dos <b style="color:#2ECC40">multiplicadores de Lagrange ou correlatos</b>. Busca-se juntar os dois requisitos acima em uma única função objeto a ser minimizada $\Psi$.



### **2.1 Método de Lagrange**

Primeiramente criamos um fator multiplicativo da condição que permita o encaixe com a função objeto anterior $\Phi$:

$$BV - W  = 0 \rightarrow 2K^T(BV-W) = 0$$
$$ \Psi =  V^TPV - 2K^T(BV-W)$$

onde $K^T = [k_1 \ \ ... \ \ k_n]$ são os chamados **multiplicadores de Lagrange** ou **correlatos**, o escalar $2$, é opcional, escolhido por conveniência algébrica como será visto mais adiante.

Agora reduzimos nosso problema em uma única minimização:
$$ \argmin_{V}(\Psi) \rightarrow \frac{\partial {\Psi}}{\partial{V}} = 0 \ \  \text{ e } \ \  \frac{\partial {\Psi}}{\partial{K}}  = 0 $$

### **2.2 Derivação com o correlato**

A derivação com relação ao resíduo nos leva a:

$$\frac{\partial {\Psi}}{\partial{V}}​ = \frac{\partial \ V^TPV}{\partial{V}} - \frac{\partial \ 2K^T(BV-W)}{\partial{V}}= 2PV - 2B^TK = 0$$

Dividindo por $2$ e isolando $V$:

$$V = P^{-1} B^T K  \rightarrow (I) $$

Já a derivada com relação ao correlato, nos leva à própria condição:

$$\frac{\partial {\Psi}}{\partial{K}}​ = \frac{\partial \ V^TPV}{\partial{K}} - \frac{\partial \ 2K^T(BV-W)}{\partial{K}}= -2BV + 2W = 0  \rightarrow (II)$$

Estamos diante de um sistema de 2 equações matriciais com 2 incógnitas, $V$ e $K$.

$$\begin{bmatrix}P & B^T \\ B & 0 \end{bmatrix}\begin{bmatrix} V \\ -K \end{bmatrix} + \begin{bmatrix} 0 \\ W \end{bmatrix} = \begin{bmatrix}0 \\ 0 \end{bmatrix}$$

### **2.2 Derivação com o correlato**

$$\begin{bmatrix}P & B^T \\ B & 0 \end{bmatrix}\begin{bmatrix} V \\ -K \end{bmatrix} + \begin{bmatrix} 0 \\ W \end{bmatrix} = \begin{bmatrix}0 \\ 0 \end{bmatrix}$$

Substituindo equação $(I)$ em $(II)$ temos:

$$B(\underbrace{P^{-1}B^TK}_{V})-W=0$$

$$(BP^{-1}B^T)K-W=0$$

$$K = (BP^{-1}B^T)^{-1} W$$

A solução para os resíduos fica:

$$V =  P^{-1} B^T (BP^{-1}B^T)^{-1} W$$

### **2.3 Propagação dos erros das observações**

$$\hat{L} =  L + V = L + P^{-1} B^T (BP^{-1}B^T)^{-1} W$$

$$\hat{L} =   L + \underbrace{P^{-1} B^T (BP^{-1}B^T)^{-1}}_{R} (C-BL)$$

$$\hat{L} =   (I-RB)L + RC$$

Pela lei de propagação das variâncias:

$$\Sigma_{\hat{L}} = (I-RB)\Sigma_L (I-RB)^T$$

Simplificando:

$$\Sigma_{\hat{L}} = \Sigma_L - \Sigma_L B^T(B\Sigma_LB^T)^{-1}B\Sigma_L $$

### **2.4 Solução do método condicionado para o caso linear**

**Resíduos:** $V =  P^{-1} B^T (BP^{-1}B^T)^{-1} W$

**Observações ajustadas:** $\hat{L} = L+V$

**Variância da observação a posteriori:** $\hat{\sigma}^2_0 = \frac{V^TPV}{m}$

**MVC das observações ajustadas:** $\Sigma_{\hat{L}} = \Sigma_L - \Sigma_L B^T(B\Sigma_LB^T)^{-1}B\Sigma_L $


**Exercício 01 (resolvido)** Em uma praça em forma de triângulo foram medidos os três ângulos A, B e C. Ajuste os ângulos respeitando a condição geométrica:

| Estação | Observado |
| ------- | --------- |
| A       |  68°10'40"|
| B       |  66°34'22"|
| C       |  45°14'49"|


**Exercício 01 (solução)**

````python
import sympy as sp
from IPython.display import display, Markdown, Math

sp.init_printing()

def dms_to_degrees(graus, minutos, segundos):
    return graus + minutos / 60 + segundos / 3600

# Observações em segundos de arco
labels = ["A", "B", "C"]
L = sp.Matrix([
    dms_to_degrees(68, 10, 40),
    dms_to_degrees(66, 34, 22),
    dms_to_degrees(45, 14, 49),
])
V = sp.Matrix(['v_a', 'v_b', 'v_c'])

print("1) Vetor das observações brutas")
display(Math(r"L=" + sp.latex(L) + r"\;\text{ graus}"))

# Condição geométrica: A + B + C = 180°
B = sp.Matrix([[1, 1, 1]])
C = sp.Matrix([dms_to_degrees(180, 0, 0)])
P = sp.eye(3)

print("2) Matrizes do modelo condicionado")
display(Math(r"F(\hat{L})=B \hat{L}-C=" +
               sp.latex(B) + sp.latex(L+V) + r"-" + sp.latex(C) + r"=0"))

print("3) Peso desconsiderado (P=I)")

# display(Math(r"P=" + sp.latex(P)))

# Fechamento em segundos: W = B L - C
W = B * L - C

print("4) Erro de fechamento")
display(Math(r"W=B L-C=" + sp.latex(W) + r"\;\text{ graus}"))
display(Markdown(f"Soma observada: **{((B * L)[0])}**. Erro de fechamento: **{(W[0])}**."))
````

````python
# Solução por correlatos de Lagrange
Pinv = P.inv()
M = B * Pinv * B.T
K = -M.inv() * W
V = Pinv * B.T * K
La = L + V

print(("5) Correlato"))
display(Math(r"K=-(BP^{-1}B^T)^{-1}W=" + sp.latex(K)))

print(("6) Resíduos"))
display(Math(r"V=P^{-1}B^TK=" + sp.latex(V) + r"\;\text{ graus}"))

print(("7) Observações ajustadas"))
display(Math(r"\hat{L}=L+V=" + sp.latex(La) + r"\;\text{ graus}"))

print(("8) Verificação da condição"))
F_La = B * La - C
display(Math(r"F(\hat{L})=B \hat{L}-C=" + sp.latex(F_La)))
display(Markdown(f"Soma ajustada: **{(B * La)[0]}**."))

sigma0_2 = (V.T * P * V)[0] / B.rows
print(("9) Variância da unidade de peso a posteriori"))
display(Math(r"\hat{\sigma}_0^2=\frac{V^TPV}{m}=" + sp.latex(sigma0_2) + r"\;\text{graus}^2"))
````

### **2.5 Solução do método condicionado para o caso não linear**

$$
F(\hat{L}^i) \cong \underbrace{F(\hat{L}^{i-1})}_{W_i \text{ ? Vai mudar...}} +
\underbrace{\left. \frac{\partial F}{\partial \hat{L}^i} \right|_{\hat{L}^{i-1}}}_{B_i}
\left( \hat{L}^i - \hat{L}^{i-1} \right)
=
F(\hat{L}^{i-1}) + B_i \left( \hat{L}^i - \hat{L}^{i-1} \right) = 0
$$

$$
\text{mas, } \quad
\hat{L}^i - \hat{L}^{i-1}
=
\hat{L}^i - L + L - \hat{L}^{i-1}
=
V_i + L - \hat{L}^{i-1}
$$

$$
\text{assim, } \quad
F(\hat{L}^i) \cong
F(\hat{L}^{i-1}) + B_i V_i + B_i \left( L - \hat{L}^{i-1} \right)
=
$$

$$
B_i V_i + \underbrace{B_i \left( L - \hat{L}^{i-1} \right) + F(\hat{L}^{i-1})}_{W_i \text{ ? Mudou...}} = 0
$$

$$\text{pelo método dos correlatos, } \quad V_i =  P^{-1} B_i^T (_iBP^{-1}B_i^T)^{-1} W_i$$

$$\text{para reiniciar o ciclo, } \quad \hat{L}^i = L + V_i $$

Até a convergência ou até algum critério de parada.

### **2.5 Solução do método condicionado para o caso não linear**

Notar que na primeira iteração:

- $L = \hat{L}^{i-1}$, o que anula o primeiro termo em $W_i = B_i( L - \hat{L}^{i-1} ) + F(\hat{L}^{i-1})$. Logo, este é o $W_i$ "genérico";
- $B_1 = \left. \frac{\partial F}{\partial \hat{L}^1} \right|_{L}$

Note a vantagem do modelo condicionado: **não é necessário um "chute" inicial** $X_0$ para parâmetros pois o ajuste atua diretamente sobre as observações.
