## 1. Interpretar a experiência de Bradley

**James Bradley** (1693–1762) foi um astrônomo inglês que, ao tentar medir a paralaxe estelar (o deslocamento aparente das estrelas causado pelo movimento da Terra), fez medições repetidas e sistemáticas da posição de uma mesma estrela ao longo do tempo. Ao registrar centenas de medições, Bradley percebeu que:

- Os resultados variavam aleatoriamente em <b style="color:#2ECC40">torno de um valor central</b>
- Os erros não eram sistemáticos, pois não havia tendência constante para um lado
- Ao calcular a média de todas as medições, o resultado era muito mais confiável do que qualquer medição isolada
- A média das medições converge para um valor
- Quanto mais medições, menor a dispersão da média, ou seja, maior confiabilidade

**Exercício 01 (resolvido):** Dado o sistema de equações redundante abaixo, prove que a média é também a solução que minimiza a soma dos quadrados dos resíduos.

$$
 \begin{cases}
m = l_1 \\
\vdots \\
m = l_n
\end{cases} 
$$

**Solução exercício 01:** Em forma matricial $A\vec{x} = \vec{l}$

$
\underbrace{\begin{bmatrix}
1 \\ \vdots \\ 1
\end{bmatrix}}_{A} \underbrace{[m]}_{\vec{x}} = 
\underbrace{\begin{bmatrix}
l_1 \\  \vdots \\ l_n
\end{bmatrix}}_{\vec{l}}
$

Aplicando a fórmula da solução MMQ: $\vec{x_a} = (A^T A)^{-1} A^T \vec{l}$

$
\vec{x_a} = \left( \begin{bmatrix}1 & \dots & 1\end{bmatrix} \begin{bmatrix}1 \\ \vdots \\ 1 \end{bmatrix} \right)^{-1} \begin{bmatrix}1 & \dots & 1\end{bmatrix} 
\begin{bmatrix}
l_1 \\  \vdots \\ l_n
\end{bmatrix} \rightarrow
$

$
\vec{x_a} = ( n )^{-1} (l_1 +  \dots + l_n) =\frac{1}{n} \sum_{i=1}^{n} l_i
$

## 2. Conceituar e diferenciar os diversos tipos de erros

Como visto anteriormente, os erros nas observações podem ser originados por **falhas humanas**, **imperfeição do equipamento**, **condições ambientais** e **efeitos aleatórios** sem causa específica. Derivados dessas naturezas, os erros são classificados como:

- **Erros Grosseiros**: grandes, detectáveis e podem ser totalmente eliminados
- **Erros Sistemáticos**: seguem lei conhecida, podendo ser reduzidos ou eliminados
- **Erros Acidentais (aleatórios)**: pequenos, imprevisíveis, distribuição normal (base do MMQ)

Perceba que **o método MMQ é uma etapa de um processo maior**. Dada uma observação $l$, se soubéssemos seu valor real $l_r$ teríamos que descontar todos os tipos de erro:

$$l = l_r - e_g - e_s - e_a$$

Onde $e_g$, $e_s$ e $e_a$ são os erros grosseiros, sistemáticos e aleatórios.

## 3. Diferenciar os conceitos de incerteza e exatidão

- **Precisão (ou incerteza)**: dispersão (desvio padrão). O grau de precisão obtido é dependente da qualidade do equipamento com que se faz as medidas, da estabilidade ambiental e da habilidade do operador em controlar os procedimentos de medição
- **Acurácia (ou exatidão)**: proximidade do valor verdadeiro. Normalmente expressa pela diferença absoluta entre uma observação e seu valor verdadeiro.
